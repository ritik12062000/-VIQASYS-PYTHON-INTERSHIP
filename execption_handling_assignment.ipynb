```python
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
    except TypeError:
        print("Error: Invalid input type")
    else:
        print("Result:", result)
    finally:
        print("Execution completed")
divide_numbers(10, 2)
divide_numbers(10, 0)
```

```python
def convert_to_integer(value):
    try:
        converted_value = int(value)
    except ValueError:
        print(f"Error: Cannot convert '{value}' to an integer due to invalid value.")
    except TypeError:
        print(f"Error: Cannot convert '{value}' to an integer due to invalid type.")
    else:
        print(f"Successfully converted to integer: {converted_value}")
    finally:
        print("Conversion attempt finished.")


convert_to_integer("123")
convert_to_integer(45.67)
```

```python
def calculate_square_of_float(value):
    try:
        float_value = float(value)
    except ValueError:
        print(f"Error: Cannot convert '{value}' to a float.")
        return None
    else:
        square = float_value ** 2
        return square
    finally:
        print("Process done.")


result1 = calculate_square_of_float(5)
print(f"Square of 5: {result1}")

result2 = calculate_square_of_float("3.14")
print(f"Square of '3.14': {result2}")

```

```python
def floor_divide_numbers(a, b):
    try:
        quotient = a // b
    except ZeroDivisionError:
        print("Error: Cannot perform floor division by zero.")
    except TypeError:
        print("Error: Invalid input types for division.")
    else:
        print(f"The floor division quotient is: {quotient}")
    finally:
        print("Function executed.")

floor_divide_numbers(10, 3)
floor_divide_numbers(10, 0)
```

```python
def find_reciprocal(number):
    try:
        reciprocal = 1 / number
    except ZeroDivisionError:
        print("Error: Cannot find reciprocal of zero.")
    except TypeError:
        print(f"Error: Invalid input type for '{number}'. Please provide a number.")
    else:
        print(f"The reciprocal of {number} is: {reciprocal}")
    finally:
        print("End of operation.")


find_reciprocal(5)
find_reciprocal(0)
find_reciprocal(2.5)
find_reciprocal("hello")
```

```python
def multiply_as_integers(a, b):
    try:
        int_a = int(a)
        int_b = int(b)
        product = int_a * int_b
    except ValueError:
        print(f"Error: Cannot convert one or both inputs ('{a}', '{b}') to integers due to invalid value.")
    except TypeError:
        print(f"Error: Invalid input types for multiplication. Inputs were '{type(a).__name__}' and '{type(b).__name__}'.")
    else:
        print(f"The product of {int_a} and {int_b} is: {product}")
    finally:
        print("Multiplication tried.")
multiply_as_integers(5, 10)
multiply_as_integers("7", "3")
multiply_as_integers(2, 4)
```

```python
def calculate_percentage(marks, total_marks):
    try:
        if not isinstance(marks, (int, float)) or not isinstance(total_marks, (int, float)):
            raise TypeError("Marks and total marks must be numbers.")
        if total_marks == 0:
            raise ZeroDivisionError("Total marks cannot be zero.")
        percentage = (marks / total_marks) * 100
    except ZeroDivisionError as e:
        print(f"Error: {e}")
    except TypeError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"Percentage: {percentage:.2f}%")
    finally:
        print("Calculation finished.")

calculate_percentage(80, 100)
calculate_percentage(75, 90)
calculate_percentage(50, 0)
calculate_percentage(90, "100")
```

```python
def convert_to_float_and_half(input_string):
    try:
        float_value = float(input_string)
    except ValueError:
        print(f"Error: Cannot convert '{input_string}' to a float.")
    else:
        half_value = float_value / 2
        print(f"Half of {float_value} is: {half_value}")
    finally:
        print("Conversion and calculation attempt finished.")
convert_to_float_and_half("10.5")
convert_to_float_and_half("20")
convert_to_float_and_half("-7.8")
```

```python
def perform_modulus(a, b):
    try:
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Both inputs must be numbers.")
        if b == 0:
            raise ZeroDivisionError("Cannot perform modulus by zero.")
        remainder = a % b
    except ZeroDivisionError as e:
        print(f"Error: {e}")
    except TypeError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"The remainder of {a} % {b} is: {remainder}")
    finally:
        print("Modulus attempt done.")
perform_modulus(10, 3)
perform_modulus(15, 4)
perform_modulus(10, 0)
```

```python
def calculate_power(base, exponent):
    try:
        num_base = float(base)
        num_exponent = float(exponent)
        result = num_base ** num_exponent
    except (ValueError, TypeError) as e:
        print(f"Error: Invalid input. Both base and exponent must be numeric. Details: {e}")
    else:
        print(f"The result of {base} raised to the power of {exponent} is: {result}")
    finally:
        print("Power function executed."
        )
calculate_power("abc", 2)
calculate_power(4, "xyz")
calculate_power(None, 3)
```

```python
def get_absolute_value(value):
    try:
        # Attempt to convert the value to a float to handle various numeric types
        numeric_value = float(value)
        absolute_val = abs(numeric_value)
    except ValueError:
        print(f"Error: Cannot convert '{value}' to a number for absolute value calculation.")
    except TypeError:
        print(f"Error: Invalid input type for '{value}'. Please provide a numeric value or something convertible to a number.")
    else:
        print(f"The absolute value of {value} is: {absolute_val}")
    finally:
        print("Absolute check completed.")

get_absolute_value(10)
get_absolute_value(-5.5)
get_absolute_value("20")
get_absolute_value("-15.3")
```

```python
def divide_three_numbers(a, b, c):
    try:
        # Convert inputs to float to handle various numeric types and prevent integer division in Python 2
        num_a = float(a)
        num_b = float(b)
        num_c = float(c)

        # Step 1: Divide a by b
        result1 = num_a / num_b
        print(f"Step 1: {a} / {b} = {result1}")

        # Step 2: Divide result1 by c
        final_result = result1 / num_c

    except ZeroDivisionError:
        print("Error: Cannot divide by zero at one of the steps.")
    except TypeError:
        print("Error: Invalid input type. Please provide numeric values.")
    except ValueError:
        print("Error: Invalid input value. Cannot convert to a number.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"Final Result of {a} / {b} / {c} is: {final_result}")
    finally:
        print("Division process ended.")

divide_three_numbers(100, 5, 2)
divide_three_numbers(50, 0, 2)
divide_three_numbers(20, 4, 0)
```

```python
def convert_and_double(value):
    try:
        # Attempt to convert the value to an integer
        int_value = int(value)
        doubled_value = int_value * 2
    except ValueError:
        print(f"Error: Cannot convert '{value}' to an integer. Invalid value provided.")
    except TypeError:
        print(f"Error: Cannot convert '{value}' to an integer. Invalid type provided.")
    else:
        print(f"The doubled integer value is: {doubled_value}")
    finally:
        print("Process completed.")

convert_and_double(10)
convert_and_double("5")
```

```python
def subtract_numbers(a, b):
    try:
        # Attempt to convert both inputs to floats for safe arithmetic
        num_a = float(a)
        num_b = float(b)
        difference = num_a - num_b
    except ValueError:
        print(f"Error: Cannot convert one or both inputs ('{a}', '{b}') to numbers. Invalid value provided.")
    except TypeError:
        print(f"Error: Invalid input type for one or both inputs ('{type(a).__name__}', '{type(b).__name__}'). Please provide numeric values.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"The difference between {a} and {b} is: {difference}")
    finally:
        print("Subtraction attempt finished.")
subtract_numbers(10, 5)
subtract_numbers(7.5, 2.3)
subtract_numbers("20", "10")
```

```python
def calculate_average(a, b):
    try:
        num_a = float(a)
        num_b = float(b)
        average = (num_a + num_b) / 2
    except ValueError:
        print(f"Error: Cannot convert one or both inputs ('{a}', '{b}') to numbers. Invalid value provided.")
    except TypeError:
        print(f"Error: Invalid input type for one or both inputs ('{type(a).__name__}', '{type(b).__name__}'). Please provide numeric values.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"The average of {a} and {b} is: {average}")
    finally:
        print("Average calculation done.")

calculate_average(10, 20)
calculate_average(7.5, 2.5)
calculate_average("50", "100")
```

```python
import math

def calculate_square_root(value):
    try:
        # Attempt to convert the value to a float
        num_value = float(value)
        if num_value < 0:
            raise ValueError("Cannot calculate square root of a negative number.")

        result = math.sqrt(num_value)
    except ValueError as e:
        print(f"Error: Invalid input for square root calculation. Details: {e}")
    except TypeError as e:
        print(f"Error: Invalid input type for '{value}'. Please provide a numeric value. Details: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    else:
        print(f"The square root of {value} is: {result}")
    finally:
        print("Square root operation finished.")
calculate_square_root(25)
calculate_square_root(10.24)
calculate_square_root("-9")
```
