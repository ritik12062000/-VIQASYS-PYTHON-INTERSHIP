````python

# 1. Multiple Inheritance: Father, Mother, and Child skills
class Father:
    def skills_father(self):
        return "Father's skills: Gardening, Carpentry"

class Mother:
    def skills_mother(self):
        return "Mother's skills: Cooking, Painting"

class Child(Father, Mother):
    def display_skills(self):
        print(self.skills_father())
        print(self.skills_mother())

# Create an instance of Child and display skills
print("--- Question 1: Multiple Inheritance ---")
child_obj = Child()
child_obj.display_skills()
````

````python
# 5. Multiple Inheritance: Animal, Bird, and Ostrich (Diamond Problem)

class Animal:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am an animal named {self.name}."

    def speak(self):
        return "(Animal sound)"

class Bird(Animal):
    def __init__(self, name, feather_color):
        super().__init__(name)
        self.feather_color = feather_color

    def introduce(self):
        base_intro = super().introduce()
        return f"{base_intro} I have {self.feather_color} feathers."

    def fly(self):
        return "I can fly!"

class Ostrich(Bird):
    def __init__(self, name, feather_color, speed):
        super().__init__(name, feather_color)
        self.speed = speed

    # Ostrich cannot fly, so override the fly method
    def fly(self):
        return "I am an Ostrich. I cannot fly, but I can run very fast!"

    def introduce(self):
        base_intro = super().introduce()
        return f"{base_intro} My top speed is {self.speed} km/h."

# Create an instance of Ostrich and demonstrate functionalities
print("--- Question 5: Multiple Inheritance (Animal, Bird, Ostrich - Diamond Problem demonstration) ---")
ostrich_obj = Ostrich("Ozzy", "brown", 70)
print(ostrich_obj.introduce())
print(ostrich_obj.speak())
print(ostrich_obj.fly())
````

````python
class Parent1:
    def common_method(self):
        return "Method from Parent1"

class Parent2:
    def common_method(self):
        return "Method from Parent2"

class Child(Parent1, Parent2):
    def common_method(self):
        # Overriding the common_method from both parents
        # You can also explicitly call parent methods if needed:
        # print(super().common_method()) # This would call Parent1's method due to MRO
        # print(Parent2.common_method(self)) # Explicitly call Parent2's method
        return "Method from Child (overridden)"

# Create an instance of Child and call the common method
print("--- Question 6: Method Overriding in Multiple Inheritance ---")
child_obj = Child()
print(child_obj.common_method())

# To demonstrate which parent's method is called by super() within Child if not overridden
class ChildWithSuper(Parent1, Parent2):
    def show_parent_method(self):
        return super().common_method() # Calls Parent1's method due to MRO

child_super_obj = ChildWithSuper()
print("Calling super().common_method() from ChildWithSuper:", child_super_obj.show_parent_method())

# To explicitly call a specific parent's method
print("Explicitly calling Parent2.common_method:", Parent2.common_method(child_obj))
````

````python
# 7. Multilevel Inheritance: Student -> GraduateStudent -> PhDStudent
class Student:
    def __init__(self, student_id, name):
        self.student_id = student_id
        self.name = name

    def get_student_info(self):
        return f"Student ID: {self.student_id}, Name: {self.name}"

class GraduateStudent(Student):
    def __init__(self, student_id, name, program, supervisor):
        super().__init__(student_id, name)
        self.program = program
        self.supervisor = supervisor

    def get_graduate_info(self):
        base_info = self.get_student_info()
        return f"{base_info}, Program: {self.program}, Supervisor: {self.supervisor}"

class PhDStudent(GraduateStudent):
    def __init__(self, student_id, name, program, supervisor, thesis_topic, graduation_year=None):
        super().__init__(student_id, name, program, supervisor)
        self.thesis_topic = thesis_topic
        self.graduation_year = graduation_year

    def get_phd_info(self):
        base_info = self.get_graduate_info()
        grad_year_info = f", Graduation Year: {self.graduation_year}" if self.graduation_year else ""
        return f"{base_info}, Thesis Topic: {self.thesis_topic}{grad_year_info}"

# Create instances and display information
print("--- Question 7: Multilevel Inheritance (Student -> GraduateStudent -> PhDStudent) ---")

s1 = Student("S001", "Alice Wonderland")
print("Student Info:", s1.get_student_info())

gs1 = GraduateStudent("G001", "Bob The Builder", "Computer Science", "Dr. Smith")
print("Graduate Student Info:", gs1.get_graduate_info())
````

````python
```python
# 9. Create two classes Person and Employee with attributes. Inherit both into Manager and display combined details.
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_person_details(self):
        return f"Name: {self.name}, Age: {self.age}"

class Employee:
    def __init__(self, employee_id, salary):
        self.employee_id = employee_id
        self.salary = salary

    def get_employee_details(self):
        return f"Employee ID: {self.employee_id}, Salary: ${self.salary:,}"

class Manager(Person, Employee):
    def __init__(self, name, age, employee_id, salary, department):
        Person.__init__(self, name, age)
        Employee.__init__(self, employee_id, salary)
        self.department = department

    def display_combined_details(self):
        print(f"Manager Details:\n  {self.get_person_details()}\n  {self.get_employee_details()}\n  Department: {self.department}")

# Create an instance of Manager and display combined details
print("--- Question 9: Person, Employee, Manager ---")
manager_obj = Manager("Alice Smith", 45, "MGR001", 120000, "Sales")
manager_obj.display_combined_details()

manager_obj_2 = Manager("Bob Johnson", 38, "MGR002", 95000, "Marketing")
manager_obj_2.display_combined_details()
```
````
