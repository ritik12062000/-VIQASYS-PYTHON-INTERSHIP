```python
def reverse_number(n):
  rev = 0
  temp = n
  while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp //= 10
  return rev
number_to_reverse = 321
reversed_number = reverse_number(number_to_reverse)
print(f"The reverse of {number_to_reverse} is {reversed_number}")
```

```python
def sum_digits(n):
  total_sum = 0
  temp = n
  while temp > 0:
    digit = temp % 10
    total_sum += digit
    temp //= 10
  return total_sum

print(sum_digits(145))
```

```python
def is_prime(n):
  if n <= 1:
    return False
  for i in range(2, n):
    if n % i == 0:
      return False
  return True

print(f"Is 7 prime? {is_prime(7)}")
print(f"Is 4 prime? {is_prime(4)}")
print(f"Is 1 prime? {is_prime(1)}")
print(f"Is 2 prime? {is_prime(2)}")
print(f"Is 11 prime? {is_prime(11)}")
print(f"Is 9 prime? {is_prime(9)}")
```

```python
def product_of_two_numbers(num1, num2):
  product = num1 * num2
  return product

print(f"The product of 5 and 7 is: {product_of_two_numbers(5, 7)}")
print(f"The product of 10 and 3 is: {product_of_two_numbers(10, 3)}")
```

```python
def remainder_of_division(dividend, divisor):
  remainder = dividend % divisor
  return remainder

print(f"The remainder of 10 divided by 3 is: {remainder_of_division(10, 3)}")
print(f"The remainder of 7 divided by 2 is: {remainder_of_division(7, 2)}")
print(f"The remainder of 15 divided by 5 is: {remainder_of_division(15, 5)}")
```

```python
def sum_up_to_number(n):
  total_sum = 0
  for i in range(1, n + 1):
    total_sum += i
  return total_sum

print(f"The sum of numbers up to 5 is: {sum_up_to_number(5)}")
print(f"The sum of numbers up to 10 is: {sum_up_to_number(10)}")
print(f"The sum of numbers up to 1 is: {sum_up_to_number(1)}")
```

```python
def generate_multiplication_table(n):
  for i in range(1, 11):
    product = n * i
    print(f"{n} x {i} = {product}")

print("Multiplication table for 5:")
generate_multiplication_table(5)
print("\nMultiplication table for 7:")
generate_multiplication_table(7)
```

```python
def calculate_power(base, exponent):
  return base ** exponent

print(f"2 raised to the power of 3 is: {calculate_power(2, 3)}")
print(f"5 raised to the power of 0 is: {calculate_power(5, 0)}")
print(f"10 raised to the power of 2 is: {calculate_power(10, 2)}")
print(f"3 raised to the power of 4 is: {calculate_power(3, 4)}")
```
