```python
# 1. Create a class called Student
class Student:
    school_name = "Springfield High"

    def __init__(self, name, age):
        self.name = name
        self.age = age

# Create three objects
student1 = Student("Alice", 16)
student2 = Student("Bob", 17)
student3 = Student("Charlie", 16)

# Print both the class attribute and instance attributes for each object
print("--- Student Class ---")
print(f"Student 1: Name={student1.name}, Age={student1.age}, School={Student.school_name}")
print(f"Student 2: Name={student2.name}, Age={student2.age}, School={student2.school_name}") # Can access via instance as well
print(f"Student 3: Name={student3.name}, Age={student3.age}, School={Student.school_name}")
```

```python
# 2. Create a class called Car
class Car:
    wheels = 4

    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

# Create two objects
car1 = Car("Toyota", 25000)
car2 = Car("Honda", 22000)

# Show how both objects share the same class attribute but have different instance attributes
print("\n--- Car Class ---")
print(f"Car 1: Brand={car1.brand}, Price={car1.price}, Wheels={Car.wheels}")
print(f"Car 2: Brand={car2.brand}, Price={car2.price}, Wheels={car2.wheels}") # Can access via instance as well

# Demonstrate shared class attribute
print(f"Are wheels attribute same for car1 and car2? {car1.wheels == car2.wheels}")
```

```python
# 3. Create a class called Employee
class Employee:
    company_name = "Tech Solutions Inc."

    def __init__(self, emp_name, salary):
        self.emp_name = emp_name
        self.salary = salary

# Create multiple employee objects
emp1 = Employee("David", 60000)
emp2 = Employee("Eve", 75000)
emp3 = Employee("Frank", 68000)

# Print their details along with the common company name
print("\n--- Employee Class ---")
print(f"Employee 1: Name={emp1.emp_name}, Salary={emp1.salary}, Company={Employee.company_name}")
print(f"Employee 2: Name={emp2.emp_name}, Salary={emp2.salary}, Company={emp2.company_name}")
print(f"Employee 3: Name={emp3.emp_name}, Salary={emp3.salary}, Company={Employee.company_name}")
```

```python
# 4. Create a class called Mobile
class Mobile:
    category = "Electronics"

    def __init__(self, mobile_name, ram):
        self.mobile_name = mobile_name
        self.ram = ram

# Create three objects
mob1 = Mobile("iPhone 13", "6GB")
mob2 = Mobile("Samsung Galaxy S21", "8GB")
mob3 = Mobile("Google Pixel 6", "8GB")

# Print all values to clearly understand class vs instance attributes
print("\n--- Mobile Class ---")
print(f"Mobile 1: Name={mob1.mobile_name}, RAM={mob1.ram}, Category={Mobile.category}")
print(f"Mobile 2: Name={mob2.mobile_name}, RAM={mob2.ram}, Category={mob2.category}")
print(f"Mobile 3: Name={mob3.mobile_name}, RAM={mob3.ram}, Category={Mobile.category}")
```

```python
# 5. Create a class called Book
class Book:
    library_name = "City Central Library"

    def __init__(self, title, author):
        self.title = title
        self.author = author

# Create at least two objects
book1 = Book("The Great Gatsby", "F. Scott Fitzgerald")
book2 = Book("1984", "George Orwell")

# Print their complete information
print("\n--- Book Class ---")
print(f"Book 1: Title='{book1.title}', Author={book1.author}, Library={Book.library_name}")
print(f"Book 2: Title='{book2.title}', Author={book2.author}, Library={book2.library_name}")
```

```python
# 6. Create a class called Laptop
class Laptop:
    warranty_years = 1

    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

# Create multiple laptop objects
lap1 = Laptop("Dell", 1200)
lap2 = Laptop("HP", 1000)
lap3 = Laptop("Lenovo", 950)

# Display how the class attribute remains same for all objects
print("\n--- Laptop Class ---")
print(f"Laptop 1: Brand={lap1.brand}, Price=${lap1.price}, Warranty={Laptop.warranty_years} years")
print(f"Laptop 2: Brand={lap2.brand}, Price=${lap2.price}, Warranty={lap2.warranty_years} years")
print(f"Laptop 3: Brand={lap3.brand}, Price=${lap3.price}, Warranty={Laptop.warranty_years} years")

print(f"All laptops have a {Laptop.warranty_years}-year warranty.")
```

```python
# 7. Create a class called Person
class Person:
    country = "USA"

    def __init__(self, name, age):
        self.name = name
        self.age = age

# Create three person objects
person1 = Person("Grace", 28)
person2 = Person("Henry", 35)
person3 = Person("Ivy", 22)

# Print their data
print("\n--- Person Class ---")
print(f"Person 1: Name={person1.name}, Age={person1.age}, Country={Person.country}")
print(f"Person 2: Name={person2.name}, Age={person2.age}, Country={person2.country}")
print(f"Person 3: Name={person3.name}, Age={person3.age}, Country={Person.country}")
```

```python
# 8. Create a class called Bike
class Bike:
    type = "Two Wheeler"

    def __init__(self, bike_name, mileage):
        self.bike_name = bike_name
        self.mileage = mileage

# Create objects
bike1 = Bike("Royal Enfield Classic", "35 kmpl")
bike2 = Bike("Bajaj Pulsar", "45 kmpl")

# Print all details
print("\n--- Bike Class ---")
print(f"Bike 1: Name={bike1.bike_name}, Mileage={bike1.mileage}, Type={Bike.type}")
print(f"Bike 2: Name={bike2.bike_name}, Mileage={bike2.mileage}, Type={bike2.type}")
```

```python
# 9. Create a class called Movie
class Movie:
    industry = "Bollywood"

    def __init__(self, movie_name, rating):
        self.movie_name = movie_name
        self.rating = rating

# Create multiple movie objects
movie1 = Movie("Dangal", 8.4)
movie2 = Movie("Lagaan", 8.1)
movie3 = Movie("3 Idiots", 8.4)

# Print the details
print("\n--- Movie Class ---")
print(f"Movie 1: Name='{movie1.movie_name}', Rating={movie1.rating}, Industry={Movie.industry}")
print(f"Movie 2: Name='{movie2.movie_name}', Rating={movie2.rating}, Industry={movie2.industry}")
print(f"Movie 3: Name='{movie3.movie_name}', Rating={movie3.rating}, Industry={Movie.industry}")
```

```python
# 10. Create a class called BankAccount
class BankAccount:
    bank_name = "Global Bank Co."

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

# Create two account objects
account1 = BankAccount("John Doe", 1500.50)
account2 = BankAccount("Jane Smith", 3200.75)

# Print their information showing both class and instance attributes
print("\n--- BankAccount Class ---")
print(f"Account 1: Holder={account1.account_holder}, Balance=${account1.balance:.2f}, Bank={BankAccount.bank_name}")
print(f"Account 2: Holder={account2.account_holder}, Balance=${account2.balance:.2f}, Bank={account2.bank_name}")
```
