```python
def largest_number(n):
  largest_digit = 0
  temp_n = abs(n)

  if temp_n == 0:
    return 0

  while temp_n > 0:
    digit = temp_n % 10
    if digit > largest_digit:
      largest_digit = digit
    temp_n //= 10
  return largest_digit
```

```python
# Write a function that takes a list of integers and returns a new list where each
# element is replaced by the product of all other elements except itself.
# Example: Input [1,2,3,4] Output [24,12,8,6]

def product_of_others(input_list):
    result = []
    for i in range(len(input_list)):
        current_product = 1
        for j in range(len(input_list)):
            if i != j:
                current_product *= input_list[j]
        result.append(current_product)
    return result
my_list = [1, 2, 3, 4]
output_list = product_of_others(my_list)
print(f"Input: {my_list}, Output: {output_list}")

```

```python
#Create a decorator that prints 'Function execution started' before execution and
# 'Function execution ended' after execution of any function. Demonstrate it on a
# function that prints numbers from 1 to n.

def my_decorator(func):
    def wrapper(*args, **kwargs):
        print('Function execution started')
        result = func(*args, **kwargs)
        print('Function execution ended')
        return result
    return wrapper

@my_decorator
def print_numbers(n):
    print(f"Printing numbers from 1 to {n}:")
    for i in range(1, n + 1):
        print(i)

# Demonstrate the decorator
print_numbers(3)
```

```python
def is_palindrome_recursive(n):
  # Handle negative numbers (typically not considered palindromes without specific definition)
  # and single-digit numbers (which are always palindromes).
  if n < 0:
    return False
  if n >= 0 and n < 10:
    return True

  # Find the divisor to extract the most significant digit (MSD)
  divisor = 1
  temp = n
  while temp >= 10:
    divisor *= 10
    temp //= 10

  msd = n // divisor
  lsd = n % 10

  if msd != lsd:
    return False
  else:
    # Remove the MSD and LSD, then recursively check the remaining number
    remaining_num = (n % divisor) // 10
    return is_palindrome_recursive(remaining_num)
print(f"Is 121 a palindrome? {is_palindrome_recursive(121)}")
print(f"Is 12321 a palindrome? {is_palindrome_recursive(12321)}")
print(f"Is 123 a palindrome? {is_palindrome_recursive(123)}")
```
