```python
class Person:
    def __init__(self):
        self.__name = ""

    def set_name(self, name):
        self.__name = name

    def get_name(self):
        return self.__name
```

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount

    def get_balance(self):
        return self.__balance
```

```python
class Student:
    def __init__(self):
        self.__marks = 0

    def set_marks(self, marks):
        if marks > 0:
            self.__marks = marks

    def get_marks(self):
        return self.__marks
```

```python
class Employee:
    def __init__(self):
        self.__salary = 0

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary

    def increase_salary(self, percent):
        if percent > 0:
            self.__salary += self.__salary * percent / 100

    def get_salary(self):
        return self.__salary
```

```python
class Temperature:
    def __init__(self):
        self.__celsius = 0

    def set_temperature(self, celsius):
        self.__celsius = celsius

    def to_fahrenheit(self):
        return (self.__celsius * 9/5) + 32
```

```python
class Mobile:
    def __init__(self):
        self.__price = 0

    def set_price(self, price):
        if price >= 0:
            self.__price = price

    def get_price(self):
        return self.__price
```

```python
class Car:
    def __init__(self):
        self.__speed = 0

    def set_speed(self, speed):
        if 0 <= speed <= 200:
            self.__speed = speed

    def get_speed(self):
        return self.__speed
```

```python
class LoginSystem:
    def __init__(self):
        self.__password = ""

    def set_password(self, password):
        self.__password = password

    def validate(self, password):
        return self.__password == password
```

```python
class Product:
    def __init__(self):
        self.__quantity = 0

    def add_stock(self, amount):
        if amount > 0:
            self.__quantity += amount

    def reduce_stock(self, amount):
        if 0 < amount <= self.__quantity:
            self.__quantity -= amount

    def get_quantity(self):
        return self.__quantity
```

```python
class VotingSystem:
    def __init__(self):
        self.__age = 0

    def set_age(self, age):
        if age > 0:
            self.__age = age

    def can_vote(self):
        return self.__age >= 18
```
